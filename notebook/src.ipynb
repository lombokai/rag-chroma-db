```python
import os
import re
import json

from rag.vector_db import ChromaDB
from rag.retriever import HybridSearch
from rag.vectorizer import Embedder
```

# Data Preparation

```python
data_path = "data/mami_bubu"

data = []
for i, filename in enumerate(os.listdir(data_path)):
    if filename.endswith(".json"):
        with open(os.path.join(data_path, filename), "r") as file:
            text = json.load(file)

        data.append({
            "id": str(i),
            "filename": filename,
            "text": str(text)
        })
```

```python
data
```

```python
def _cleaning(text: str):
    text = text.lower()

    # remove punctuation
    text = re.sub(r"[^\w\s]", "", text)
    
    return text
```

```python
for d in data:
    d["text"] = _cleaning(d["text"])

data
```

# vector db

```python
db = ChromaDB(collection_name="rag-collection")
```

## vectorizer

```python
embedder = Embedder(
    embedder_path = "weights/mxbai-embeddings",
    dimensions = 512
)
```

```python
for i, d in enumerate(data):
    d["embedding"] = embedder.embed(d["text"])
```

## insert embeddings

```python
for d in data:
    db.insert_vectors(
        ids = [d["id"]],
        docs = [d["text"]],
        embeddings = [d["embedding"]]
    )
```

# Hybrid Search

```python
sample_query = "aufa baby set"
```

```python
search = HybridSearch(
    data = data,
    embedder_path = "weights/mxbai-embeddings",
    collection_name = "rag-collection",
    db = ChromaDB,
    embedder = Embedder
)
```

```python
search.bm25_search(sample_query, top_n=10)
```

```python
search.transformer_search(sample_query)
```

```python
search.hybrid_search(
    sample_query,
    top_n=10,
    sparse_weight=0.3,
    dense_weight=0.7
)
```
